# Unemployment Analysis in Python
Data loading and understanding

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the datasets

df1 = pd.read_csv('/content/Unemployment_Rate_upto_11_2020.csv')
df2 = pd.read_csv('/content/Unemployment in India.csv')

print("Dataset 1 loaded successfully!")
print("Dataset 2 loaded successfully!")


#Dataset understanding
print("Dataset 1 - First 5 Rows")
display(df1.head())

print("\nDataset 2 - First 5 Rows")
display(df2.head())


#Check the no.of rows and columns
print("Dataset 1 Shape:", df1.shape)
print("Dataset 2 Shape:", df2.shape)


#Check column names
print("Dataset 1 Columns:")
print(df1.columns.tolist())

print("\nDataset 2 Columns:")
print(df2.columns.tolist())


#Check complete dataset information
print("Dataset 1 Information:")
df1.info()

print("\n" + "="*60 + "\n")

print("Dataset 2 Information:")
df2.info()


#Check statistical summary
print("Dataset 1 Statistical Summary:")
display(df1.describe(include='all'))

print("\nDataset 2 Statistical Summary:")
display(df2.describe(include='all'))


#Check missing values
print("Missing Values in Dataset 1:")
print(df1.isnull().sum())

print("\nMissing Values in Dataset 2:")
print(df2.isnull().sum())


#Check duplicate records
print("Duplicate rows in Dataset 1:", df1.duplicated().sum())
print("Duplicate rows in Dataset 2:", df2.duplicated().sum())
```

Data cleaning and Preprocessing

```python
# Remove leading/trailing spaces from column names

df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

print("Dataset 1 Columns after cleaning:")
print(df1.columns.tolist())

print("\nDataset 2 Columns after cleaning:")
print(df2.columns.tolist())

#Check missing values
print("Missing values in Dataset 1:")
display(df1.isnull().sum())

print("\nMissing values in Dataset 2:")
display(df2.isnull().sum())

# Check rows containing missing values in Dataset 2

missing_rows = df2[df2.isnull().any(axis=1)]

print("Number of rows containing missing values:", len(missing_rows))

display(missing_rows)

#Check Duplicate Records
print("Duplicate rows in Dataset 1:", df1.duplicated().sum())
print("Duplicate rows in Dataset 2:", df2.duplicated().sum())

#Check actual date value
print("Dataset 1 dates:")
print(df1['Date'].head(10).tolist())

print("\nDataset 2 dates:")
print(df2['Date'].head(10).tolist())

#Check categorical values
print("Dataset 1 Regions:")
print(df1['Region'].unique())

print("\nDataset 2 Regions:")
print(df2['Region'].unique())

print("Dataset 2 Areas:")
print(df2['Area'].unique())
```

Remove duplicates and missing values

```python
# Check the duplicate records in Dataset 2

duplicate_rows = df2[df2.duplicated(keep=False)]

print("Number of duplicate records:", len(duplicate_rows))

display(
    duplicate_rows.sort_values(
        ["Region", "Date", "Area"]
    )
)

# Check whether the duplicate rows are completely identical

print(
    "Exact duplicate rows:",
    df2.duplicated().sum()
)

# Show the duplicated Region-Date-Area combinations

duplicate_combinations = (
    df2.groupby(
        ["Region", "Date", "Area"],
        dropna=False
    )
    .size()
    .reset_index(name="Count")
)

display(
    duplicate_combinations[
        duplicate_combinations["Count"] > 1
    ]
)

# Remove completely empty rows from Dataset 2

df2 = df2.dropna(how="all").copy()
# Remove exact duplicate records

df2 = df2.drop_duplicates().copy()

print("Dataset 2 shape after cleaning:", df2.shape)

print("\nMissing values:")
print(df2.isnull().sum())

print("\nDuplicates:")
print(df2.duplicated().sum())

print(df1["Date"].dtype)
print(df2["Date"].dtype)
```

Data Validation and Comparison

```python
import pandas as pd

# Convert 'Date' columns to datetime objects and extract 'Year'
df1['Date'] = pd.to_datetime(df1['Date'], dayfirst=True)
df2['Date'] = pd.to_datetime(df2['Date'], dayfirst=True)

df1['Year'] = df1['Date'].dt.year
df2['Year'] = df2['Date'].dt.year

#Check the date ranges
print("Dataset 1 date range:")
print(df1["Date"].min(), "to", df1["Date"].max())

print("\nDataset 2 date range:")
print(df2["Date"].min(), "to", df2["Date"].max())

#Check the number of observations by year
print("\nDataset 1 observations by year:")
print(df1["Year"].value_counts().sort_index())

print("\nDataset 2 observations by year:")
print(df2["Year"].value_counts().sort_index())

#Check regions
print("Dataset 1 number of regions:",
      df1["Region"].nunique())

print("Dataset 2 number of regions:",
      df2["Region"].nunique())

#Check Rural/Urban distribution
print("Dataset 2 Area distribution:")
print(df2["Area"].value_counts())

#Check the common regions
common_regions = set(df1["Region"]).intersection(
    set(df2["Region"])
)

print("Number of common regions:", len(common_regions))
print("\nCommon regions:")
print(sorted(common_regions))

#Check numerical columns
numeric_columns = [
    "Estimated Unemployment Rate (%)",
    "Estimated Employed",
    "Estimated Labour Participation Rate (%)"
]

print("Dataset 1:")
display(df1[numeric_columns].describe())

print("\nDataset 2:")
display(df2[numeric_columns].describe())

print("Dataset 1 negative values:")
print((df1[numeric_columns] < 0).sum())

print("\nDataset 2 negative values:")
print((df2[numeric_columns] < 0).sum())

#Check unemployment rate
print("Dataset 1 unemployment rate range:")
print(
    df1["Estimated Unemployment Rate (%)"].min(),
    "to",
    df1["Estimated Unemployment Rate (%)"].max()
)

print("\nDataset 2 unemployment rate range:")
print(
    df2["Estimated Unemployment Rate (%)"].min(),
    "to",
    df2["Estimated Unemployment Rate (%)"].max()
)
```

Calculate unemployment before and during COVID

```python
# Calculate average unemployment rate for each period

covid_unemployment = (
    df2.groupby("Period")["Estimated Unemployment Rate (%)"]
    .mean()
)

print("Average Unemployment Rate:")
print(covid_unemployment)
```

Calculate the COVID impact

```python
# Get the average unemployment rates

pre_covid = covid_unemployment["Pre-COVID"]
covid_period = covid_unemployment["COVID Period"]

# Calculate the change in percentage points
change = covid_period - pre_covid

# Calculate percentage increase
percentage_increase = (change / pre_covid) * 100

print(f"Pre-COVID average unemployment rate: {pre_covid:.2f}%")
print(f"COVID average unemployment rate: {covid_period:.2f}%")
print(f"Increase in unemployment rate: {change:.2f} percentage points")
print(f"Percentage increase: {percentage_increase:.2f}%")
```

Visualize the COVID impact

```python
# Visualize Pre-COVID vs COVID unemployment

plt.figure(figsize=(8, 5))

covid_unemployment.plot(
    kind="bar"
)

plt.title("Average Unemployment Rate: Pre-COVID vs COVID")
plt.xlabel("Period")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=0)

plt.show()
```

Monthly unemployment trend graph



```python
# Calculate monthly average unemployment rate

monthly_unemployment = (
    df2.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

print(monthly_unemployment)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_unemployment["Date"],
    monthly_unemployment["Estimated Unemployment Rate (%)"],
    marker="o"
)

plt.title("Monthly Unemployment Rate Trend")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

Calculate the peak and recovery

```python
# Find the month with the highest unemployment rate

peak_month = monthly_unemployment.loc[
    monthly_unemployment["Estimated Unemployment Rate (%)"].idxmax()
]

print("Peak unemployment month:")
print(peak_month)

# Calculate decline from the COVID peak to June 2020

may_rate = monthly_unemployment.loc[
    monthly_unemployment["Date"] == "2020-05-31",
    "Estimated Unemployment Rate (%)"
].iloc[0]

june_rate = monthly_unemployment.loc[
    monthly_unemployment["Date"] == "2020-06-30",
    "Estimated Unemployment Rate (%)"
].iloc[0]

decline = may_rate - june_rate

print(f"May 2020 unemployment rate: {may_rate:.2f}%")
print(f"June 2020 unemployment rate: {june_rate:.2f}%")
print(f"Decline from May to June: {decline:.2f} percentage points")
```

Regional COVID impact

```python
# Calculate average unemployment by region and COVID period

region_period = (
    df2.groupby(["Region", "Period"])
    ["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

display(region_period.head(10))
```

Calculate regional increase

```python
# Reshape the data so each period becomes a separate column

regional_comparison = region_period.pivot(
    index="Region",
    columns="Period",
    values="Estimated Unemployment Rate (%)"
)

# Keep only regions having both periods
regional_comparison = regional_comparison.dropna(
    subset=["Pre-COVID", "COVID Period"]
)

# Calculate the increase in percentage points
regional_comparison["Increase (percentage points)"] = (
    regional_comparison["COVID Period"]
    - regional_comparison["Pre-COVID"]
)

# Sort from highest increase to lowest
regional_comparison = regional_comparison.sort_values(
    "Increase (percentage points)",
    ascending=False
)

display(regional_comparison)
```

Visualize the top affected regions

```python
# Select top 10 regions with the largest increase

top_10_regions = (
    regional_comparison
    .nlargest(10, "Increase (percentage points)")
    .sort_values("Increase (percentage points)")
)

# Plot
plt.figure(figsize=(10, 6))

plt.barh(
    top_10_regions.index,
    top_10_regions["Increase (percentage points)"]
)

plt.title("Top 10 Regions with Highest Increase in Unemployment")
plt.xlabel("Increase in Unemployment Rate (Percentage Points)")
plt.ylabel("Region")

plt.show()
```

Rural vs Urban COVID impact and comparison

```python
# Calculate unemployment rate by Area and COVID period

area_covid = (
    df2.groupby(["Area", "Period"])
    ["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

display(area_covid)

# Reshape the data for comparison
area_comparison = area_covid.pivot(
    index="Area",
    columns="Period",
    values="Estimated Unemployment Rate (%)"
)

area_comparison["Increase (percentage points)"] = (
    area_comparison["COVID Period"]
    - area_comparison["Pre-COVID"]
)

display(area_comparison)
```

# Policy Recommendations and Implications

### 1. Support Employment Recovery During Economic Crises

The unemployment rate increased from 9.51% before COVID-19 to 17.77% during the COVID-19 period. This indicates the significant impact of the pandemic on employment. Government and policymakers should strengthen emergency employment-support programs during major economic disruptions.

### 2. Focus on Highly Affected Regions

Puducherry recorded the highest increase in unemployment, at 37.36 percentage points. This shows that the impact of an economic crisis can vary significantly across regions. Employment-support policies should therefore consider regional differences and provide additional assistance to highly affected areas.

### 3. Strengthen Urban Employment Support

Urban unemployment increased by 8.43 percentage points during the COVID period, slightly higher than the rural increase of 8.09 percentage points. This suggests that urban workers were also strongly affected by the economic disruption. Policies should support urban employment through job-creation programs, support for small businesses, and skill-development initiatives.

### 4. Strengthen Rural Employment Opportunities

Although the rural increase was slightly lower than the urban increase, rural unemployment still increased substantially. Existing rural employment programs and livelihood-support schemes should be strengthened during economic crises to protect vulnerable workers.

### 5. Improve Employment Monitoring and Early-Warning Systems

The unemployment rate reached a peak of 24.88% in May 2020. Regular monitoring of unemployment indicators can help policymakers identify sudden changes in the labour market and respond more quickly to future economic disruptions.

### 6. Promote Skill Development and Reskilling

Economic disruptions can change the demand for different types of jobs. Skill-development and reskilling programs can help unemployed workers transition into sectors with better employment opportunities.

### 7. Strengthen Support for Vulnerable Workers

The large increase in unemployment highlights the need for targeted support for workers who are more vulnerable during economic crises. Temporary financial assistance, employment programs, and access to skill-development opportunities can help reduce the impact of sudden job losses.

## Conclusion

This analysis examined unemployment trends in India and evaluated the impact of the COVID-19 pandemic on employment conditions. The findings show that the average unemployment rate increased from 9.51% before COVID-19 to 17.77% during the COVID-19 period, representing an increase of 8.26 percentage points. The unemployment rate reached its highest observed level of 24.88% in May 2020 before declining in June.

The analysis also revealed significant regional differences in the impact of COVID-19, with Puducherry recording the highest increase in unemployment at 37.36 percentage points. Urban unemployment increased by 8.43 percentage points, while rural unemployment increased by 8.09 percentage points, showing that both areas were substantially affected.

Furthermore, the strong negative correlation of -0.951 between estimated employment and unemployment indicates that higher unemployment was closely associated with lower employment levels in the dataset. Overall, the analysis highlights the significant effect of the COVID-19 pandemic on India's labour market and emphasizes the importance of timely employment support, regional intervention, skill development, and effective labour-market monitoring during future economic disruptions.
